1.- Dada la sucesión $a_{n} y = 49,8 + 2,5_{n}$.

a) Determine $a_{10}$

b) Determine n para cuando $a_{n} = 312,3$

```python
# Respuesta a)

sucesion = []
for i in range(10):
    n = i + 1
    sucesion.append(49.8 + 2.5*n)
    print(f'El término {n} es {sucesion[i]}')
```

```python
# Respuesta b)

final = 312.3 
termino = 0
n = 0
dif = final - termino
while dif > 0:
    n += 1
    termino = 49.8 + 2.5*n
    dif = final - termino

if dif == 0:
    print(f'El {final} es un término de la sucesión y ocupa el lugar {n}')

else:
    print(f'El {final} no pertenece a la sucesión')
```

2.- Dada la sucesión $2n^{4}$ - $5n^{3}$ - $2n^{2}$.
